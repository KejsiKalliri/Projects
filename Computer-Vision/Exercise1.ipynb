```python
import numpy as np

img = img_as_float(data.coffee())

noisy_image_1 = random_noise(img,mode='gaussian')
local_variance = 0.01 * np.ones(img.shape)
noisy_image_2 = random_noise(img,mode='localvar', local_vars = local_variance)
noisy_image_3 = random_noise(img, mode='poisson')
noisy_image_4 = random_noise(img,mode='salt')
noisy_image_5 = random_noise(img,mode='pepper')
noisy_image_6 = random_noise(img,mode='s&p')
noisy_image_7 = random_noise(img,mode='speckle')

fig = plt.figure(figsize=(14,16))

plt.subplot(4,2,1)
plt.imshow(noisy_image_1,cmap='gray')
plt.title('gaussian')
plt.xticks([])
plt.yticks([])

plt.subplot(4,2,2)
plt.imshow(noisy_image_2,cmap='gray')
plt.title('localvar')
plt.xticks([])
plt.yticks([])

plt.subplot(4,2,3)
plt.imshow(noisy_image_3,cmap='gray')
plt.title('poisson')
plt.xticks([])
plt.yticks([])

plt.subplot(4,2,4)
plt.imshow(noisy_image_4,cmap='gray')
plt.title('salt')
plt.xticks([])
plt.yticks([])

plt.subplot(4,2,5)
plt.imshow(noisy_image_5,cmap='gray')
plt.title('pepper')
plt.xticks([])
plt.yticks([])

plt.subplot(4,2,6)
plt.imshow(noisy_image_6,cmap='gray')
plt.title('s&p')
plt.xticks([])
plt.yticks([])

plt.subplot(4,2,7)
plt.imshow(noisy_image_7,cmap='gray')
plt.title('speckle')
plt.xticks([])
plt.yticks([])

plt.show()
```

```python
#3 different intensities
noisy_image_11 = random_noise(img,mode='gaussian',var=0.005)
noisy_image_12 = random_noise(img,mode='gaussian',var=0.02)
noisy_image_13 = random_noise(img,mode='gaussian',var=0.08)


fig = plt.figure(figsize=(14,10))

plt.subplot(2,2,1)
plt.imshow(noisy_image_1,cmap='gray')
plt.title('original')
plt.xticks([])
plt.yticks([])

plt.subplot(2,2,2)
plt.imshow(noisy_image_11,cmap='gray')
plt.title('gaussian variance=0.005')
plt.xticks([])
plt.yticks([])

plt.subplot(2,2,3)
plt.imshow(noisy_image_12,cmap='gray')
plt.title('gaussian variance=0.02')
plt.xticks([])
plt.yticks([])

plt.subplot(2,2,4)
plt.imshow(noisy_image_13,cmap='gray')
plt.title('gaussian variance=0.08')
plt.xticks([])
plt.yticks([])

plt.show()
```

```python
from skimage.filters import gaussian
from skimage.color import rgb2gray

img_f1 = gaussian(noisy_image_11,0.5)
img_f2 = gaussian(noisy_image_11,2)
img_f3 = gaussian(noisy_image_11,5)

fig = plt.figure(figsize=(14,10))

plt.subplot(2,2,1)
plt.imshow(noisy_image_11,cmap='gray')
plt.title('original')
plt.xticks([])
plt.yticks([])

plt.subplot(2,2,2)
plt.imshow(img_f1,cmap='gray')
plt.title('sigma = 0.5')
plt.xticks([])
plt.yticks([])

plt.subplot(2,2,3)
plt.imshow(img_f2,cmap='gray')
plt.title('sigma = 2')
plt.xticks([])
plt.yticks([])

plt.subplot(2,2,4)
plt.imshow(img_f3,cmap='gray')
plt.title('sigma = 5')
plt.xticks([])
plt.yticks([])

plt.show()

#COMMENTS :

# Small σ (0.5) only slightly smooths—good when you want to preserve almost all detail but need minimal noise reduction.

# Intermediate σ (2) generally offers the best noise‑vs‑detail trade‑off: it removes visible speckle while keeping contours and textures recognizable.

# Large σ (5) drastically reduces noise at the cost of blurring edges and textures—useful if you only care about very large structures and can sacrifice fine detail.

# By tuning σ, you control the scale of smoothing: low σ acts on high‑frequency noise, high σ also removes mid‑frequency image content.
#Choose σ based on the size of the features you wish to preserve versus the strength of the noise you need to eliminate.
```

```python
from skimage import filters
from skimage.feature import canny

img = img_as_float(data.brick())  #edge detection algorithms take normalized images (values [0,1]) or 8-bit integer with values [0,255]

edge_prewitt = filters.prewitt(img)
#edge_roberts = filters.roberts(img)
#edge_sobel = filters.sobel(img)
#edge_canny = canny(img, sigma=1)

edge_mask = edge_prewitt > 0.05

#COMMENT :

# It simply performs an element-wise comparison on the edge_response matrix (which holds the value of the Prewitt filter at each pixel)
# and creates a Boolean matrix (edge_mask) with the same dimensions:

# Every location where edge_response[i,j] is greater than 0.05 -> edge_mask[i,j] = True

# Every location where edge_response[i,j] is less than or equal to 0.05 -> edge_mask[i,j] = False

total = img.size

edges_count = edge_mask.sum()   #all 1's are summed , which represent edges

bricks_count = total - edges_count

print(f"Total pixels:          {total}")
print(f"Edge pixels:           {edges_count} ({edges_count/total:.2%})")
print(f"Bridge pixels:       {bricks_count} ({bricks_count/total:.2%})")


fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(img, cmap='gray')
axes[0].set_title("Original Bricks")
axes[0].axis('off')

axes[1].imshow(edge_prewitt, cmap='gray')
axes[1].set_title("Prewitt Response")
axes[1].axis('off')

axes[2].imshow(edge_mask,cmap='gray')
axes[2].set_title(f"Edge Mask\n({edges_count/total:.2%} of pixels)")
axes[2].axis('off')

plt.show()
```

```python
image = data.grass()

fig = plt.figure(figsize=(14,7))
plt.imshow(image)
plt.show()

image.shape
```

```python
img = img_as_float(image)

f = np.fft.fft2(img) # Discrete Fourier Transform using FFT algorithm
fshift = np.fft.fftshift(f) # Shift to have zero frequency (origin) at the center of the image
magnitude_spectrum = 20*np.log(np.abs(fshift)) # Logarithmic scale for better visualization of the differences in power spectrum

fig = plt.figure(figsize=(14, 6))

plt.subplot(121)
plt.imshow(img, cmap = 'gray')
plt.title('Input Image')
plt.xticks([])
plt.yticks([])

plt.subplot(122)
plt.imshow(magnitude_spectrum, cmap = 'gray')
plt.title('Magnitude Spectrum')
plt.xticks([])
plt.yticks([])

plt.savefig('frequency.png')

plt.show()

# The original image (grass) has a dense texture, meaning it contains many high-frequency components.
```

```python
rows, cols = img.shape
crows , ccols = int(rows/2) , int(cols/2)
fshift[crows-30:crows+30 , ccols-30:ccols+30] = 0

f_ishift = np.fft.ifftshift(fshift)
img_back = np.fft.ifft2(f_ishift)
img_back = np.abs(img_back)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
axes[0].imshow(img, cmap='gray')
axes[0].set_title("Original")
axes[0].axis('off')

axes[1].imshow(img_back, cmap='gray')
axes[1].set_title("High-pass filter")
axes[1].axis('off')

# High-pass filter (HPF):
# It allows only high frequencies to pass, which means it preserves fine details.
# The result highlights sharp contours, but the overall structure (low frequencies) disappears.
# It resembles the application of an edge detector.
```

```python
nrows, ncols = img.shape
row , col = np.ogrid[:nrows, :ncols]

cnt_row , cnt_col = nrows/2 , ncols/2
disk_radius = 50
outer_disk_mask = ((row-cnt_row)**2 + (col-cnt_col)**2 > (disk_radius)**2)

fshift[outer_disk_mask] = 0
f_shift = np.fft.ifftshift(fshift)
img_back = np.fft.ifft2(f_shift)
img_back = np.abs(img_back)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
axes[0].imshow(img, cmap='gray')
axes[0].set_title("Original")
axes[0].axis('off')

axes[1].imshow(img_back, cmap='gray')
axes[1].set_title("Low-pass filter")
axes[1].axis('off')


# Low-pass filter (LPF):
# It allows only low frequencies to pass through.
# The result is a smoothed image where fine details are lost, but the overall shape is preserved.
# It removes details such as leaves or grass fibers.
```

```python
from skimage import data

img_blob = data.binary_blobs()

fig = plt.figure(figsize=(14,7))

plt.subplot(1,2,1)
plt.imshow(img_blob)
plt.xticks([])
plt.yticks([])

plt.subplot(1,2,2)
plt.imshow(img_blob,cmap='gray')
plt.xticks([])
plt.yticks([])

plt.show()
```

```python
from skimage.morphology import disk
from skimage.morphology import erosion

original = img_blob.copy()

footprint = disk(3)
counter = 0;

while img_blob.any():
  img_blob = erosion(img_blob,footprint)
  counter+=1

fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].imshow(original, cmap='gray')
ax[0].set_title('Original binary blobs')

ax[1].imshow(img_blob, cmap='gray')
ax[1].set_title(f'After {counter} erosions')

ax[0].axis('off')
ax[1].axis('off')

plt.tight_layout()
plt.show()
```
